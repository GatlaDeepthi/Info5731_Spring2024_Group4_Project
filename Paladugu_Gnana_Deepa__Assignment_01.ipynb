```python
from string import punctuation
import re

# (a)
# taking inputs from user for sentance 1, sentance 2
sentence1 =input("Sentence 1: ")
sentence2 =input("Sentence 2: ")


# (b)
def sentence_splitting(sentence):
  # splitting the sentence into words
  split_words=sentence.split()
  output=[]
  for word in split_words:
    # splitting the special characters from words
    if word[-1] in ".,?;'-#@!:":
      output.append(word[:-1])
      output.append(word[-1])
    else:
      output.append(word)
  return output
split_list=sentence_splitting(sentence1)+sentence_splitting(sentence2)
print(split_list)


# (c)

common_words=[]
for word in sentence_splitting(sentence1):
  # comparing two sentence and finding common words
  if word.lower() in (word.lower() for word in sentence_splitting(sentence2)):
    common_words.append(word)
print("Common words: ",common_words)


# (d)
def  sanitize_sentene(sentence):
  punctuations = ".;':,?{}()[]"
  # removing punctuations
  for word in punctuations:
    sentence = sentence.replace(word, "")
  return sentence
print("sanitized sentence1: ",sanitize_sentene(sentence1))
print("sanitized sentence2: ",sanitize_sentene(sentence2))


# (e)
def vowels_count(sentence):
  vowels="aeiou"
  count={}
  # counting the occurence of vovels in sentence
  for letter in vowels:
    count[letter]=sentence.lower().count(letter)
  return count
print("vovels count in sentence1: ",vowels_count(sentence1))
print("vovels count in sentence1: ",vowels_count(sentence2))






```

Design and implement a simple command-line calculator program. The program should handle both integers and floating-point numbers.

(d) Implement error handling in the calculator program. Handle cases such as division by zero and invalid input expressions. Provide meaningful error messages.

(e) Implement a command to exit the calculator.

```python
import math

def find_expression(expression):
  # splitting the expression
  exp=expression.split()
  # if len is 2 it can be square or exponential and followed by number
  if len(exp)==2:
    ex,num1=exp
    num1=float(num1)
    if ex=="sqrt":
      if num1<0:
        return "Error: can't calculate sqrt for negaative numbers"
      return math.sqrt(num1)
    elif ex=="exp":
        return math.exp(num1)
    else:
        return "Error: invalid operator"

# if len is 3 it can be number followed by expression and followed by another number
  if len(exp)==3:
    num1,ex,num2=exp
    num2=float(num2)
    num1=float(num1)
    if ex=="+":
      return num1+num2
    elif ex=="-":
      return num1-num2
    elif ex=="/":
      if num2==0:
        return "Error: Invalid input, Can't perform division by zero"
      return num1/num2
    elif ex=="*":
      return num1*num2
    elif ex=="%":
      if num2==0:
        return "Error: Invalid input, Can't perform modulo by zero"
      return num1%num2
    elif ex=="**":
      return num1**num2
    elif ex == "sqrt":
      if num1 < 0:
          return "Error: Invalid input,Can't perform  square root for negative number."
      return math.sqrt(num1)
    elif ex == "exp":
      return math.exp(num1)
    else:
      return "Invalid operator"
  else:
    return "Invalid expression, required format: number operator number (e.g., 5 * 7)"

def calculator(expression):
  # checking foe valid paranthesis
  try:
    while '(' in expression:
      last_open_bracket=expression.rfind('(')
      last_close_bracket=expression.find(')',last_open_bracket)
      if last_close_bracket==-1:
        return "Error: Invalid paranthesis"
      sub_expression=expression[last_open_bracket+1:last_close_bracket]
      sub_result=str(eval(sub_expression))
      expression=expression[:last_open_bracket]+sub_result+expression[last_close_bracket+1:]
    return find_expression(expression)
  except Exception as e:
    return e

print("Use 'quit' to exit the calculator")
while True:
  expression=input("Enter yuour expression: ")
  # quitting from the calculater by using 'quit'
  if expression.lower()=="quit":
    break
  print(calculator(expression))

```

# Question 5 (30 points).

In the field of Data Science, data is often formatted as a comma-delimited (CSV) file, where each line in a file corresponds to a field's value. Refer to Chapter 9 (p. 106) for more information on comma-delimited files.

To answer this question, you will need to download the file [Assignment1_denton_housing.csv](https://raw.githubusercontent.com/unt-iialab/info5731-spring2022/main/assignments/Assignment1_denton_housing.csv) included with this assignment. The file contains statistics about housing in the city of Denton from 2008 to 2014. Write a program to implement the following questions.

(a) (5 pts) Calculate the difference in the number of Occupied Housing Units from year to year and print it. The difference must be calculated for the consecutive years such as 2008-2009, 2009-2010 etc. Finally, print the values in the ascending order.

(b) (5 pts) For all the years, calculate the percentage of housing units which are vacant an occupied. Print the results in the following format:

Year Vacant Housing Units Occupied Housing Units

    2008  30%   70%                          
	2009 ----- -----
	2010 ----- -----
    2011 ----- -----
	2012 ----- -----
    2013 ----- -----
	2014 ----- -----
  
  

(c) (10 pts) Calculate and print the valued and years in which the highest number of housing units were vacant and occupied. Print the results in the following format:  

                              Year	Value
	Vacant Housing Units	  -----   -----
	Occupied Housing Units	-----   -----

(d) (10 pts) Calculate the harmonic mean of the total housing units and print it out. You can find the information about harmonic mean here: https://ncalculators.com/statistics/harmonic-mean-calculator.htm

```python
import pandas as pd
# (a)
# loading csv
dataframe = pd.read_csv('Assignment1_denton_housing.csv')
df = dataframe.pivot(index='year', columns='title_field', values='value').reset_index()
df['Difference'] = df['Occupied Housing Units'].diff()
# removing null values
df = df.dropna(subset=['Difference'])
print("Difference of Occupied Housing Units")
print(df['Difference'].sort_values(ascending=True))


# (b)
# calculating vacant percentage and occupied percentage
df = dataframe.pivot(index='year', columns='title_field', values='value').reset_index()

df['Occupied Percentage'] = (df['Occupied Housing Units'] /df['Total Housing Units']) * 100
df['Vacant Percentage'] = (df['Vacant Housing Units'] / df['Total Housing Units']) * 100

# Printing the results
for index, row in df.iterrows():
    year = int(row['year'])
    vacant_percentage = f"{row['Vacant Percentage']:.0f}%"
    occupied_percentage = f"{row['Occupied Percentage']:.0f}%"
    print(f"{year} {vacant_percentage} {occupied_percentage}")


# (c)
# calculating maximum vecant and occupied housing units
df = dataframe.pivot(index='year', columns='title_field', values='value').reset_index()

maximum_vacant_units = df.loc[df['Vacant Housing Units'].idxmax()]
maximum_occupied_units = df.loc[df['Occupied Housing Units'].idxmax()]

print(f"Vacant Housing Units      {maximum_vacant_units['year']:>5}   {maximum_vacant_units['Vacant Housing Units']:>5}")
print(f"Occupied Housing Units    {maximum_occupied_units['year']:>5}   {maximum_occupied_units['Occupied Housing Units']:>5}")


# (d)
from scipy.stats import hmean
total_units = df['Total Housing Units']

# Calculating hmean of total_units
hmean = hmean(total_units[total_units > 0])
print(f"Harmonic mean: {hmean:.2f}")
```
